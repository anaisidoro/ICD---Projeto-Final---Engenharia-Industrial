```python
import pandas as pd
import io

# Nome do arquivo carregado
file_name = 'EAF_LIST_dados_para_Colab.csv'

try:
    # Lendo o arquivo CSV (tratando o encoding UTF-8 com BOM que é comum em arquivos CSV exportados)
    df = pd.read_csv(file_name, encoding='utf-8-sig')
    print(f"Arquivo '{file_name}' carregado com sucesso!\n")
except Exception as e:
    print(f"Erro ao ler o arquivo como CSV: {e}")
    print("Tentando ler como Excel...")
    try:
        # Caso o arquivo seja na verdade um Excel com extensão alterada ou vice-versa
        df = pd.read_excel(file_name)
        print(f"Arquivo '{file_name}' carregado com sucesso como Excel!\n")
    except Exception as ex:
        print(f"Erro ao ler o arquivo: {ex}")

# 1. Quantidade total de registros
total_registros = len(df)
print(f"Quantidade total de registros: {total_registros}\n")

# 2. Nomes das colunas originais
print("Nomes das colunas originais:")
print(list(df.columns))
print("\n" + "="*50 + "\n")

# 3. Tratamento inicial de dados
# Remover colunas que estão completamente vazias
df.dropna(how='all', axis=1, inplace=True)

# Remover linhas que estão completamente vazias
df.dropna(how='all', axis=0, inplace=True)

# Remover espaços em branco extras dos nomes das colunas e dos valores de texto
df.columns = df.columns.str.strip()
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].astype(str).str.strip()

# Substituir strings vazias ou representações de 'nan' por valores nulos reais (None/NaN)
df.replace(['', 'nan', 'NaN', 'None'], pd.NA, inplace=True)

# 4. Apresentar as primeiras 10 linhas
print("Primeiras 10 linhas do DataFrame tratado:")
display(df.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Filtrar registros onde o tipo é desvio (DEVIATION)
df_deviations = df[df['TYPE'].str.upper() == 'DEVIATION'].copy()

# 2. Remover valores ausentes ou nulos do PART NUMBER para a contagem
df_deviations_clean = df_deviations.dropna(subset=['PART NUMBER'])
df_deviations_clean = df_deviations_clean[df_deviations_clean['PART NUMBER'] != '<NA>']

# 3. Agrupar por Part Number e contar as ocorrências
pn_counts = df_deviations_clean['PART NUMBER'].value_counts().reset_index()
pn_counts.columns = ['PART NUMBER', 'Ocorrências']

# 4. Selecionar os top 20 Part Numbers
top_20_pns = pn_counts.head(20)

# Exibir a tabela de ocorrências
print("=== TOP 20 PART NUMBERS COM MAIS DESVIOS ===")
display(top_20_pns)

# 5. Plotar o gráfico de barras em ordem decrescente
plt.figure(figsize=(12, 8))
sns.barplot(
    x='Ocorrências',
    y='PART NUMBER',
    data=top_20_pns,
    palette='viridis',
    hue='PART NUMBER',
    legend=False
)

plt.title('Top 20 Part Numbers com Maior Número de Desvios Registrados', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Quantidade de Ocorrências', fontsize=12)
plt.ylabel('Part Number', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Definir palavras-chave relacionadas à falta de peças/componentes
termos_falta = ['SHORTAGE', 'FALTA', 'MISSING', 'NOT FOUND', 'SEM PEÇA', 'FALTANDO']

# Criar uma máscara booleana para encontrar esses termos na coluna DESCRIPTION
# Convertendo a coluna para string e tratando nulos temporariamente para o filtro
filtro_falta = df['DESCRIPTION'].fillna('').str.upper().apply(
    lambda desc: any(termo in desc for termo in termos_falta)
)

# 2. Filtrar o DataFrame original
df_falta = df[filtro_falta].copy()

# Limpar a coluna MODEL para remover nulos ou valores inválidos
df_falta_modelos = df_falta.dropna(subset=['MODEL'])
df_falta_modelos = df_falta_modelos[df_falta_modelos['MODEL'] != '<NA>']

# 3. Agrupar por MODEL e contar as ocorrências de alterações
ranking_modelos = df_falta_modelos['MODEL'].value_counts().reset_index()
ranking_modelos.columns = ['MODELO_MAQUINA', 'Alterações por Falta de Peças']

# Exibir o ranking dos modelos mais afetados
print("=== RANKING DOS MODELOS MAIS AFETADOS POR FALTA DE PEÇAS ===")
display(ranking_modelos)

# 4. Gerar o gráfico comparativo
if not ranking_modelos.empty:
    plt.figure(figsize=(12, 6))
    sns.barplot(
        x='Alterações por Falta de Peças',
        y='MODELO_MAQUINA',
        data=ranking_modelos,
        palette='Oranges_r',
        hue='MODELO_MAQUINA',
        legend=False
    )

    plt.title('Ranking de Modelos de Máquinas Mais Afetados por Falta de Peças', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Quantidade de Alterações Registradas', fontsize=12)
    plt.ylabel('Modelo da Máquina', fontsize=12)
    plt.grid(axis='x', linestyle='--', alpha=0.7)

    plt.tight_layout()
    plt.show()
else:
    print("Nenhum registro correspondente aos filtros de falta de peças foi associado a um modelo específico de máquina.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import re

# 1. Utilizar os mesmos registros filtrados por falta de peças (df_falta)
# Vamos trabalhar diretamente com os modelos não nulos
df_modelos_brutos = df_falta.dropna(subset=['MODEL']).copy()
df_modelos_brutos = df_modelos_brutos[df_modelos_brutos['MODEL'] != '<NA>']

# 2. Separar modelos agrupados por barras '/', vírgulas ',' ou ponto e vírgula ';'
lista_modelos_individuais = []
for index, row in df_modelos_brutos.iterrows():
    model_cell = str(row['MODEL'])
    # Divide a string usando múltiplos delimitadores
    modelos_divididos = re.split(r'[/,;]', model_cell)
    for modelo in modelos_divididos:
        modelo_limpo = modelo.strip()
        if modelo_limpo and modelo_limpo != '<NA>':
            lista_modelos_individuais.append(modelo_limpo)

# 3. Criar DataFrame de contagem individualizada
df_individual = pd.DataFrame(lista_modelos_individuais, columns=['MODELO'])
ranking_individual = df_individual['MODELO'].value_counts().reset_index()
ranking_individual.columns = ['MODELO_MAQUINA', 'Alterações por Falta de Peças']

# Selecionar apenas o TOP 10
top_10_modelos = ranking_individual.head(10)

# Exibir a tabela formatada
print("=== TOP 10 MODELOS DE MÁQUINAS MAIS AFETADOS POR FALTA DE PEÇAS ===")
display(top_10_modelos)

# 4. Gerar o gráfico de barras horizontais otimizado
plt.figure(figsize=(12, 7))

# Plotagem com seaborn
ax = sns.barplot(
    x='Alterações por Falta de Peças',
    y='MODELO_MAQUINA',
    data=top_10_modelos,
    palette='Reds_r',
    hue='MODELO_MAQUINA',
    legend=False
)

# Adicionar os valores numéricos ao final de cada barra
for i in ax.patches:
    width = i.get_width()
    # Centraliza o texto verticalmente e dá um pequeno espaçamento horizontal
    ax.text(
        width + 0.5,
        i.get_y() + i.get_height() / 2,
        f'{int(width)}',
        ha='left',
        va='center',
        fontsize=11,
        fontweight='bold',
        color='black'
    )

# Customização visual do gráfico para máxima legibilidade
plt.title('Top 10 Modelos de Máquinas Mais Afetados por Falta de Peças\n(Contagem Individualizada de Modelos)', fontsize=14, fontweight='bold', pad=20)
plt.xlabel('Quantidade de Ocorrências (Alterações)', fontsize=12, labelpad=10)
plt.ylabel('Modelo da Máquina', fontsize=12, labelpad=10)

# Configurar limites do eixo X para dar margem para os textos de valor
max_valor = top_10_modelos['Alterações por Falta de Peças'].max()
plt.xlim(0, max_valor * 1.15)

# Ajustes de borda e grade
sns.despine(left=True, bottom=True)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Definir mapeamento de categorias baseado em palavras-chave frequentes nas descrições
# Isso garante um agrupamento semântico muito mais preciso e interpretável do que clusterização puramente estatística.
categorias_problemas = {
    'Falta de Peças / Componentes (Shortage)': ['SHORTAGE', 'FALTA', 'MISSING', 'NOT FOUND', 'SEM PEÇA', 'FALTANDO', 'SHORTAGE OF'],
    'Inconformidade Dimensional / Especificação': ['DIMENSION', 'SPEC', 'MEDIDA', 'FORA DE MEDIDA', 'DESVIO DIMENSIONAL', 'LARGURA', 'COMPRIMENTO', 'TOLERÂNCIA', 'OUT OF SPEC'],
    'Problemas de Montagem / Ajuste': ['ASSEMBLY', 'MONTAGEM', 'AJUSTE', 'FIT', 'NÃO ENTRA', 'DIFÍCIL ACESSO', ' interferência', 'INTERFERENCE'],
    'Defeitos de Solda / Trincas': ['WELD', 'SOLDA', 'TRINCA', 'CRACK', 'SPLAY', 'RESPINGO'],
    'Problemas de Pintura / Acabamento': ['PAINT', 'PINTURA', 'CORROSÃO', 'RUST', 'ARRANHADO', 'RISCADO', 'ACABAMENTO'],
    'Divergência de Desenho / Documentação': ['DRAWING', 'DESENHO', 'DOCUMENTAÇÃO', 'DIVERGÊNCIA', 'CAD', 'DIFERENÇA'],
    'Problemas com Furos / Rosca / Fixação': ['HOLE', 'FURO', 'ROSCA', 'THREAD', 'PARAFUSO', 'BOLT', 'ESPAÇAMENTO', 'BOSS'],
    'Tubulações / Mangueiras / Vazamentos': ['HOSE', 'PIPE', 'MANGUEIRA', 'TUBO', 'VAZAMENTO', 'LEAK', 'FLANGE'],
    'Problemas Elétricos / Chicotes': ['HARNESS', 'CHICOTE', 'ELÉTRICO', 'SENSOR', 'BATERIA', 'CABLE', 'CABO'],
    'Inconsistência de Estoque / Logística': ['STOCK', 'ESTOQUE', 'INVENTÁRIO', 'DIVERGENCIA DE ESTOQUE', 'LOGÍSTICA']
}

def categorizar_descricao(desc):
    if pd.isna(desc):
        return 'Não Informado'
    desc_upper = str(desc).upper()
    for categoria, palavras_chave in categorias_problemas.items():
        if any(palavra in desc_upper for palavra in palavras_chave):
            return categoria
    return 'Outros Desvios / Diversos'

# 2. Aplicar a categorização no DataFrame completo
df['CATEGORIA_PROBLEMA'] = df['DESCRIPTION'].apply(categorizar_descricao)

# Calcular a frequência e o percentual
tabela_problemas = df['CATEGORIA_PROBLEMA'].value_counts().reset_index()
tabela_problemas.columns = ['Categoria do Problema', 'Frequência']
tabela_problemas['Percentual (%)'] = (tabela_problemas['Frequência'] / total_registros) * 100

# Exibir os 10 principais resultados
top_10_problemas = tabela_problemas.head(10)

print("=== TABELA DE PROBLEMAS MAIS RECORRENTES NAS EAFs ===")
display(top_10_problemas.style.format({'Percentual (%)': '{:.2f}%'}))

# 3. Plotar gráfico de barras de padrão acadêmico e profissional
plt.figure(figsize=(13, 7))
sns.set_theme(style="whitegrid")

ax = sns.barplot(
    x='Frequência',
    y='Categoria do Problema',
    data=top_10_problemas,
    palette='Blues_r',
    hue='Categoria do Problema',
    legend=False
)

# Adicionar rótulos com a contagem e o percentual no final de cada barra
for i in ax.patches:
    width = i.get_width()
    percentual = (width / total_registros) * 100
    ax.text(
        width + 5,
        i.get_y() + i.get_height() / 2,
        f'{int(width)} ({percentual:.1f}%)',
        ha='left',
        va='center',
        fontsize=11,
        fontweight='semibold',
        color='#2c3e50'
    )

# Ajustes estéticos e de espaçamento
plt.title('Distribuição das Principais Categorias de Ocorrências nas EAFs', fontsize=15, fontweight='bold', pad=22, color='#2c3e50')
plt.xlabel('Quantidade de Ocorrências (Frequência Absoluta)', fontsize=12, labelpad=10, color='#2c3e50')
plt.ylabel('Categoria Identificada', fontsize=12, labelpad=10, color='#2c3e50')

# Dar margem para exibição dos textos das barras
plt.xlim(0, top_10_problemas['Frequência'].max() * 1.15)

sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Tratamento da coluna DATE
def converter_data(val):
    if pd.isna(val):
        return pd.NaT
    val_str = str(val).strip()

    # Tenta converter se for número inteiro (formato de data do Excel, ex: 43605)
    try:
        if val_str.isdigit():
            return pd.to_datetime(int(val_str), unit='D', origin='1899-12-30')
    except:
        pass

    # Tenta conversão padrão para formatos de data comuns
    return pd.to_datetime(val_str, errors='coerce')

# Criar coluna com datas tratadas
df['DATE_CLEANED'] = df['DATE'].apply(converter_data)

# Filtrar registros que possuem data válida
df_dates = df.dropna(subset=['DATE_CLEANED']).copy()

# Criar coluna Ano-Mês para agrupamento temporal
df_dates['ANO_MES'] = df_dates['DATE_CLEANED'].dt.to_period('M')

# 2. Agrupamento por Ano-Mês
evolucao_mensal = df_dates['ANO_MES'].value_counts().sort_index().reset_index()
evolucao_mensal.columns = ['Ano-Mês', 'Quantidade de EAFs']
# Converter o período de volta para string para melhor plotagem e manipulação
evolucao_mensal['Ano-Mês-Str'] = evolucao_mensal['Ano-Mês'].astype(str)

# 3. Identificar os 10 meses com maior número de registros
top_10_meses = evolucao_mensal.sort_values(by='Quantidade de EAFs', ascending=False).head(10).copy()
top_10_meses['Percentual (%)'] = (top_10_meses['Quantidade de EAFs'] / len(df_dates)) * 100

print("=== TOP 10 MESES COM MAIOR NÚMERO DE REGISTROS DE EAFs ===")
display(top_10_meses[['Ano-Mês-Str', 'Quantidade de EAFs', 'Percentual (%)']].style.format({'Percentual (%)': '{:.2f}%'}))

# 4. Gerar Gráfico de Linha da Evolução Temporal (Padrão Acadêmico)
plt.figure(figsize=(14, 6))
sns.set_theme(style="whitegrid")

# Linha de tendência temporal
ax = sns.lineplot(
    x='Ano-Mês-Str',
    y='Quantidade de EAFs',
    data=evolucao_mensal,
    marker='o',
    linewidth=2.5,
    color='#1f77b4',
    label='EAFs Abertas'
)

# Destacar os pontos dos 3 meses de maior pico para fins de análise acadêmica
picos = evolucao_mensal.sort_values(by='Quantidade de EAFs', ascending=False).head(3)
plt.scatter(
    picos['Ano-Mês-Str'],
    picos['Quantidade de EAFs'],
    color='red',
    s=100,
    zorder=5,
    label='Meses de Pico Máximo'
)

# Adicionar anotações de valores nos pontos de pico
for idx, row in picos.iterrows():
    plt.annotate(
        f"{int(row['Quantidade de EAFs'])}",
        (row['Ano-Mês-Str'], row['Quantidade de EAFs']),
        textcoords="offset points",
        xytext=(0,10),
        ha='center',
        fontsize=10,
        fontweight='bold',
        color='red'
    )

# Ajustar rotação e espaçamento do eixo X para evitar sobreposição
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)

plt.title('Evolução Mensal de Abertura de EAFs (Série Histórica)', fontsize=15, fontweight='bold', pad=20, color='#2c3e50')
plt.xlabel('Período (Ano-Mês)', fontsize=12, labelpad=10, color='#2c3e50')
plt.ylabel('Quantidade de Ocorrências', fontsize=12, labelpad=10, color='#2c3e50')

plt.legend(frameon=True, facecolor='white', edgecolor='none', fontsize=11)
sns.despine()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Limpeza dos Part Numbers
# Filtra linhas sem PN válido (evitando nulos, '<NA>' ou hífens solitários)
df_pn = df.dropna(subset=['PART NUMBER']).copy()
df_pn = df_pn[~df_pn['PART NUMBER'].isin(['<NA>', '-', '', 'None', 'SEVERAL PNS'])]

# 2. Identificar os 10 Part Numbers com maior reincidência
top_10_pns_series = df_pn['PART NUMBER'].value_counts().head(10)
top_10_pns_lista = top_10_pns_series.index.tolist()

# 3. Analisar problemas e modelos associados a esses 10 PNs
analise_pn = []

for pn in top_10_pns_lista:
    sub_df = df_pn[df_pn['PART NUMBER'] == pn]
    total_eafs = len(sub_df)

    # Obter o problema mais frequente associado a este PN
    problema_principal = sub_df['CATEGORIA_PROBLEMA'].value_counts().idxmax()

    # Obter os modelos mais afetados (unindo múltiplos se houver empate ou listando os principais)
    modelos_afetados = sub_df['MODEL'].dropna()
    modelos_afetados = modelos_afetados[modelos_afetados != '<NA>']
    if not modelos_afetados.empty:
        top_modelos = ", ".join(modelos_afetados.value_counts().head(2).index.tolist())
    else:
        top_modelos = "Não Informado"

    analise_pn.append({
        'Part Number': pn,
        'Reincidências (EAFs)': total_eafs,
        'Problema Principal': problema_principal,
        'Modelos Mais Afetados': top_modelos
    })

df_analise_pn = pd.DataFrame(analise_pn)

# Exibir a tabela analítica formatada para apresentação
print("=== TOP 10 PART NUMBERS COM MAIOR REINCIDÊNCIA E SUAS ASSOCIAÇÕES ===")
display(df_analise_pn.style.set_properties(**{'text-align': 'left'}).hide(axis='index'))

# 4. Gerar gráfico comparativo horizontal profissional
plt.figure(figsize=(13, 7))
sns.set_theme(style="whitegrid")

ax = sns.barplot(
    x='Reincidências (EAFs)',
    y='Part Number',
    data=df_analise_pn,
    palette='crest_r',
    hue='Part Number',
    legend=False
)

# Adicionar os valores numéricos de forma limpa e visível ao final de cada barra
for i in ax.patches:
    width = i.get_width()
    ax.text(
        width + 0.15,
        i.get_y() + i.get_height() / 2,
        f'{int(width)} EAFs',
        ha='left',
        va='center',
        fontsize=11,
        fontweight='bold',
        color='#2c3e50'
    )

# Customização estética padrão acadêmico
plt.title('Top 10 Part Numbers com Maior Reincidência de Desvios/Alterações', fontsize=15, fontweight='bold', pad=22, color='#2c3e50')
plt.xlabel('Quantidade de EAFs Associadas', fontsize=12, labelpad=10, color='#2c3e50')
plt.ylabel('Part Number', fontsize=12, labelpad=10, color='#2c3e50')

# Ajustar limite dinâmico do eixo X para acomodar os rótulos de dados
max_val = df_analise_pn['Reincidências (EAFs)'].max()
plt.xlim(0, max_val + 1.2)

sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Tratamento e Filtragem de Outliers no Tempo de Resposta
# Criando uma cópia para preservar os dados originais
df_analise_tempo = df.dropna(subset=['BUSINESS_DAYS_CLEANED', 'STATUS_CLEANED']).copy()

# Filtrando outliers usando o método do IQR para melhorar a qualidade visual da análise acadêmica
Q1 = df_analise_tempo['BUSINESS_DAYS_CLEANED'].quantile(0.25)
Q3 = df_analise_tempo['BUSINESS_DAYS_CLEANED'].quantile(0.75)
IQR = Q3 - Q1
limite_superior = Q3 + 1.5 * IQR

# Mantendo registros dentro do limite ou pelo menos um intervalo razoável (ex: até 180 dias se houver muita variação)
df_filtrado = df_analise_tempo[df_analise_tempo['BUSINESS_DAYS_CLEANED'] <= max(limite_superior, 30)].copy()

# 2. Agrupamento estatístico
estatisticas_status = df_filtrado.groupby('STATUS_CLEANED')['BUSINESS_DAYS_CLEANED'].agg(
    Quantidade='count',
    Média_Dias_Úteis='mean',
    Mediana_Dias_Úteis='median',
    Desvio_Padrão='std'
).reset_index()

print("=== ANÁLISE DESCRITIVA: TEMPO DE RESPOSTA VS STATUS ===")
display(estatisticas_status.style.format({
    'Média_Dias_Úteis': '{:.2f}',
    'Mediana_Dias_Úteis': '{:.2f}',
    'Desvio_Padrão': '{:.2f}'
}).hide(axis='index'))

# 3. Geração do Gráfico de Boxplot Acadêmico
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Plot comparativo de caixas
ax = sns.boxplot(
    x='STATUS_CLEANED',
    y='BUSINESS_DAYS_CLEANED',
    data=df_filtrado,
    palette='Set2',
    hue='STATUS_CLEANED',
    legend=False,
    linewidth=1.8
)

# Adicionar a média como um ponto de destaque no gráfico para comparação direta
sns.stripplot(
    x='STATUS_CLEANED',
    y='BUSINESS_DAYS_CLEANED',
    data=df_filtrado,
    color='#2c3e50',
    size=4,
    alpha=0.3,
    jitter=0.2
)

# Customizações estéticas
plt.title('Distribuição do Tempo de Resposta por Status da EAF\n(Removidos Outliers Extremos para Melhor Visualização)', fontsize=14, fontweight='bold', pad=18, color='#2c3e50')
plt.xlabel('Status da EAF', fontsize=12, labelpad=10, color='#2c3e50')
plt.ylabel('Tempo de Resposta (Dias Úteis)', fontsize=12, labelpad=10, color='#2c3e50')

sns.despine()
plt.tight_layout()
plt.show()
```
